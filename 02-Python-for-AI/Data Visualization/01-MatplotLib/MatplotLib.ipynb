```python
# Import Libirary
import matplotlib.pyplot as plt
```

```python
# 1. Line Plot
months = ['Jan','Feb','Mar','Apr']
sales = [25000,31000,65000,45000]
plt.plot(months,sales,marker='o')
plt.title("Months sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
#. Bar Chart
product = ['Laptop','Mobial','Tablet']
sales = [50,80,40]
plt.bar(product,sales,color='orange')
plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python
plt.barh(product,sales,color='pink')
plt.title("Product Sales")
plt.xlabel("Sales")
plt.ylabel("Product")
plt.show()
```

```python
# Scatter Plot
hours = [1,2,3,4,5]
marks = [35,45,55,65,75]
plt.scatter(hours,marks,color='red')
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
# Histogram
marks = [45,50,55,60]
plt.hist(marks,color='orange',bins=10)
plt.title("Marks Distrubution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
# Pie Chart
product = ['Mobial','Laptop','Tablet']
sales = [35,40,55]
plt.pie(sales,labels=product,autopct="%1.1f%%")
plt.title("Product Sales Percentage")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python
# Subplot

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [20000, 25000, 22000, 30000]
orders = [100, 120, 110, 150]

plt.subplot(1, 2, 1)

plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")

plt.subplot(1, 2, 2)

plt.bar(months, orders)
plt.title("Monthly Orders")

plt.show()
```

```python
# Question
"""
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [45000, 52000, 48000, 60000, 75000, 68000],
    "Orders": [120, 150, 130, 180, 220, 200],
    "Customers": [90, 110, 100, 140, 170, 160]
}

1.Which month has the highest sales?
2.Which month has the lowest sales?
3.Which month has the highest number of orders?
4.Is there a relationship between the number of customers and sales?
5.Which visualization is best for showing the sales trend over time, and why?
6.Which visualization is best for comparing monthly orders, and why?
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [45000, 52000, 48000, 60000, 75000, 68000],
    "Orders": [120, 150, 130, 180, 220, 200],
    "Customers": [90, 110, 100, 140, 170, 160]
}


df = pd.DataFrame(data)
df.head()
```

```python
# 1.Which month has the highest sales?
# 2.Which month has the lowest sales?
plt.bar(df['Month'],df['Sales'])
plt.title("Months Vs Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
# 3.Which month has the highest number of orders?
plt.bar(df['Month'],df['Orders'])
plt.title("Months Vs Orders")
plt.xlabel("Months")
plt.ylabel("Orders")
plt.show()
```

```python
# 4.Is there a relationship between the number of customers and sales?

plt.scatter(df['Customers'],df['Sales'])
plt.title("Customers vs Sales")
plt.ylabel("Sales")
plt.show()
```

```python
# 5.Which visualization is best for showing the sales trend over time, and why?

plt.plot(df['Month'],df['Sales'],marker='o')
plt.title("Months Vs Sales")
plt.xlabel("months")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
# 6.Which visualization is best for comparing monthly orders, and why?
plt.bar(df['Month'],df['Orders'],color='pink')
plt.title("Months Vs Orders")
plt.xlabel("Months")
plt.ylabel("Orders")
plt.show()
```

```python
plt.pie(df['Sales'],labels=df['Month'],autopct="%1.1f%%")
plt.title("Months Vs Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```
