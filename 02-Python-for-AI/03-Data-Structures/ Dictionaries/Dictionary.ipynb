```python
# Study od Dictionary

# Create Dictionary
student = {
    "Name": "Samarth",
    "Age": 33,
    "Marks": 90
}
print(student)

# 2.Acces Element
print(student["Name"])
print(student['Age'])

# 3.Add element in dictionary
student["City"] = 'Nagpur'
print(student)

# 4.Update Elemnt Value in Dict
student["Marks"] = 95
print(student)

# 5.Delete Element in Dictionary
student.pop('City')
print(student)

# 6.Element Values keys in Dictionary
print(student.keys())

# 7. Values of Dictionary
print(student.values())

# 8. Items in dictionay
print(student.items())

# 9.Uses of loop
for value in student.values():
  print(value)
for key,value in student.items():
  print(key,value)

# 10. Copy and clear function in Dictionary
student1 = student.copy()
print(student1)

student1.clear()
print(student1)
```

```python
#Practice Question 1
"""
1.Student Record Management
student = {
    "roll_no": 101,
    "name": "Samarth Kokate",
    "age": 22,
    "branch": "CSE-DS",
    "year": "Final Year",
    "cgpa": 7.50,
    "college": "SCET Nagpur",
    "city": "Nagpur",
    "email": "samarth@gmail.com",
    "phone": "9876543210"
}
Tasks:
1.Print the student's name, branch, and CGPA.
2.Update the CGPA to 7.80.
3.Change the city to "Pune".
4.Add "skills": ["Python", "SQL", "Power BI"].
5.Remove the "phone" key.
6.Print all keys.
7.Print all values.
8.Print all key-value pairs.
"""

#1.Print the student's name, branch, and CGPA.
student = {
    "roll_no": 101,
    "name": "Samarth Kokate",
    "age": 22,
    "branch": "CSE-DS",
    "year": "Final Year",
    "cgpa": 7.50,
    "college": "SCET Nagpur",
    "city": "Nagpur",
    "email": "samarth@gmail.com",
    "phone": "9876543210"
}
print(f"Student Name:{student['name']}\nBranch:{student['branch']}\nCGPA:{student['cgpa']}")
print("="*30)

#2.Update the CGPA to 7.80.
student['cgpa'] = 7.80
print(f"Student Name:{student['name']}\nBranch:{student['branch']}\nCGPA:{student['cgpa']}")
print("="*30)

#3.Change the city to "Pune".
print("After Change City Nagpur To Pune")
student['city'] = 'Pune'
print(student)
print("="*30)

#4.Add "skills": ["Python", "SQL", "Power BI"].
student['skills'] = ["Python","SQL","Power BI"]
print("Ater Add Skills")
print(student)
print("="*30)

#5.Remove the "phone" key.
student.pop('phone')
print("After Remove Phone Key")
print(student)
print("="*30)

#6.Print all keys.
print("All Kreys")
print(student.keys())
print("="*30)

#7.Print all values.
print("All Values")
print(student.values())
print("="*30)

#8.Print all key-value pairs.
print("All key and value in pairs")
print(student.items())
print("=*30")
```

```python
#Practice Question 3

"""
3.Product Catalog
product = {
    "product_id": "P1001",
    "name": "Wireless Mouse",
    "brand": "Logitech",
    "category": "Electronics",
    "price": 799,
    "discount": 10,
    "stock": 25,
    "rating": 4.5,
    "seller": "TechStore",
    "warranty": "2 Years"
}
Tasks:
1.Print the product name, price, and brand.
2.Update the stock to 18.
3.Increase Discount by 15%
4.Add "color": "Black".
5.Remove the "seller" key.
6.Print all product information.
"""

product = {
    "product_id": "P1001",
    "name": "Wireless Mouse",
    "brand": "Logitech",
    "category": "Electronics",
    "price": 799,
    "discount": 10,
    "stock": 25,
    "rating": 4.5,
    "seller": "TechStore",
    "warranty": "2 Years"
}
#1.Print the product name, price, and brand.
print(f"Product Name:{product['name']}\nPrice:{product['price']}\nBrand:{product['brand']}")
print("="*30)

#2.Update the stock to 18.
product['stock'] = 18
print("After Stock Update:",product['stock'])
print("="*30)

#3.Increase Discount by 15%
product['discount'] = product['discount'] + 15
print("After Discount Increase by 15% is =",product['discount'])
print("="*30)

#4.Add "color": "Black".
product['color'] = 'Black'
print("After Add New Color as Black")
print(product)
print("="*30)


#5.Remove the "seller" key.
product.pop('seller')
print("after Remove Seller key:",product)
print("="*30)

#6.Print all product information.
print("======Product Details=====")
for key,value in product.items():
  print(key,":",value)
print("="*30)
```

```python
#Practice Question 5

"""
5.API Response & JSON Data
api_response = {
    "status": "success",
    "status_code": 200,
    "message": "Login Successful",
    "user": {
        "id": 101,
        "name": "Samarth Kokate",
        "email": "samarth@gmail.com",
        "phone": "9876543210",
        "city": "Nagpur",
        "role": "Student"
    },
    "token": "ABC123XYZ",
    "login_time": "10:30 AM",
    "device": "Laptop",
    "ip_address": "192.168.1.20"
}
Tasks:
1.Print the user's name and email.
2.Print the login time and device.
3.Update the token to "XYZ789ABC".
4.Change the user's city to "Pune".
5.Add "branch": "CSE-DS" inside the user dictionary.
6.Print the complete updated dictionary.
"""


api_response = {
    "status": "success",
    "status_code": 200,
    "message": "Login Successful",
    "user": {
        "id": 101,
        "name": "Samarth Kokate",
        "email": "samarth@gmail.com",
        "phone": "9876543210",
        "city": "Nagpur",
        "role": "Student"
    },
    "token": "ABC123XYZ",
    "login_time": "10:30 AM",
    "device": "Laptop",
    "ip_address": "192.168.1.20"
}
#1.Print the user's name and email
print(f"User Name:{api_response['user']['name']}\nEmail:{api_response['user']['email']}")
print("="*30)

#2.Print the login time and device.
print(f"Login Time:{api_response['login_time']}\nDevice:{api_response['device']}")
print("="*30)

#3.Update the token to "XYZ789ABC".
api_response['token'] = "XYZ789ABC"
print("After Token Update:",api_response['token'])
print("="*30)

#4.Change the user's city to "Pune".
api_response['user']['city'] = "Pune"
print("After Change City To Pune:",api_response['user']['city'])
print("="*30)

#5.Add "branch": "CSE-DS" inside the user dictionary.
api_response['user']['branch'] = "CSE_DS"
print("After Add Branch :",api_response['user']['branch'])
print("="*30)

#6.Print the complete updated dictionary
print("=====Updated Dictionary=====")
for key,value in api_response.items():
  print(key,":",value)
print("="*30)
```
