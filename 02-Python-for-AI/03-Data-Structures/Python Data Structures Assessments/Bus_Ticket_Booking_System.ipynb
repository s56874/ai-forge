```python
# Bus Ticket Boking System

buses = {
    "MH101": {
        "Route": "Nagpur → Pune",
        "Seats": 40,
        "Fare": 850
    },
    "MH102": {
        "Route": "Nagpur → Mumbai",
        "Seats": 35,
        "Fare": 950
    },
    "MH103": {
        "Route": "Nagpur → Nashik",
        "Seats": 30,
        "Fare": 700
    },
    "MH104": {
        "Route": "Nagpur → Aurangabad",
        "Seats": 25,
        "Fare": 600
    }
}

bookings = [{
    "Name": "Samarth",
    "Bus": "MH101",
    "Seats": 2
}]

while True:
  print("======Welcome to Bus Ticket Booking System======")
  print("1. View Avilable Buses")
  print("2. Book Ticket")
  print("3. View Booking")
  print("4. Search Booking")
  print("5. Cancel Booking")
  print("6. Total Bookings")
  print("7. Exit")
  print("="*30)


  # Taking Input from User
  try:
        choice = int(input("\nEnter Your Choice: "))
  except ValueError:
      print("Please enter a valid number.")
      continue


# 1. To shov Avilable Buses in the System
  if choice == 1:

   print("======Avilable Bus Details======")
   for bus_number, bus_details in buses.items():
     print(f"Bus Number: {bus_number}")
     print(f"Route: {bus_details['Route']}")
     print(f"Seats: {bus_details['Seats']}")
     print(f"Fare: {bus_details['Fare']}")
     print("="*30)

  # 2. Book Ticket
  elif choice == 2:
    print("======Book Ticket======")
    bus_number = input("Enter bus Number:")
    if bus_number in buses:
      name = input("Enter Passenger Name:")
      seats = int(input("Enter Number of Seats:"))
      if seats > 0 and seats <= buses[bus_number]['Seats']:
        total_fare = seats * buses[bus_number]['Fare']
        print(f"Total Fare: {total_fare}")
        confirm = input("Do You Want Confirm Booking:(Yes/No)")
        if confirm.lower() == 'yes':
          bookings.append({
              'Name': name,
              'Bus' : bus_number,
              'Seats': seats
          })

          buses[bus_number]['Seats'] -= seats
          print("Booking Confirmed")
        else:
          print("Booking Cancelled")

      else:
        print("Invalid Number of Seats")
    else:
      print("Invalid Bus Number")



   # 3.View Booking in the system
  elif choice == 3:
    print("======View Bookings======")
    if not bookings:
      print("No Bookings found")
    else:
      for booking in bookings:
        print(f"Passenger Name: {booking['Name']}")
        print(f"Bus: {booking['Bus']}")
        print(f"Seats: {booking['Seats']}")
        print("="*30)



  # 4. Search Booking
  elif choice == 4:
    print("======Search Booking======")
    name = input("Enter Passengers Name:")
    found = False
    for booking in bookings:
      if booking["Name"] == name:
        print(f"Passenger Name: {booking['Name']}")
        print(f"Bus: {booking['Bus']}")
        print(f"Seats: {booking['Seats']}")
        print("="*30)
        found = True
      if not found:
       print("Booking Not Found")



  # 5.. Cancel Booking
  elif choice == 5:
    print("======Cancel Booking======")
    name = input("Enter Passenger Name:")
    found = False
    for booking in bookings:
      if booking['Name'] == name:
        bus_number = booking['Bus']
        seats = booking['Seats']
        bookings.remove(booking)
        buses[bus_number]['Seats'] += seats
        print("Booking Canceled")
        found = True
        break
    if not found:
      print("Booking Not Found")


  # 6.Total Bookings
  elif choice == 6:
    print("=======Total Bookings======")
    total_booking = len(bookings)
    print("Total Bokings :",total_booking)


# 7. Exit System
  elif choice == 7:
    print("===== Exiting Bus Ticket Booking System =====")
    print("Thank You for Using Our System.")
    break


  else:
    print("Invalid Input!Please Enter Valid Input ")
```
