```python
#Check missing values
df.isnull()

# Check missing values Count
df.isnull().sum()

# Remove missing values
df.dropna()
```

```python
# Question 1
"""


data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Rohan", "Priya"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "HR"],
    "Age": [25, 28, 30, np.nan, 32, 28],
    "Salary": [50000, 45000, 60000, 55000, np.nan, 45000]
}
1.Display the first 3 rows.
2.Check the shape and data types.
3.Find missing values.
4.Fill missing Age with the average age.
5.Fill missing Salary with the average salary.
6.Find duplicate rows and remove them.
7.Display employees earning more than 50,000.
8.Add a Bonus column equal to 10% of Salary.
9.Sort employees by Salary in descending order.
10.Find the average salary of each department.
"""

import numpy as np
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Rohan", "Priya"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "HR"],
    "Age": [25, 28, 30, np.nan, 32, 28],
    "Salary": [50000, 45000, 60000, 55000, np.nan, 45000]
}

df = pd.DataFrame(data)

#1.Display the first 3 rows.
df.head(5)


```

```python
#2.Check the shape and data types
df.shape
df.dtypes
```

```python
#3.Find missing values
df.isnull()
df.isnull().sum()
```

```python
#4.Fill missing Age with the average age.
df['Age'] = df['Age'].fillna(df['Age'].mean())
df.isnull().sum()
```

```python
#5..Fill missing Salary with the average Salary.
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df.isnull().sum()
```

```python
#6.Find duplicate rows and remove them.
df.duplicated()
df.drop_duplicates()
```

```python
#7.Display employees earning more than 50,000.
df[df['Salary']> 50000]
```

```python
#8.Add a Bonus column equal to 10% of Salary.
df['Bonus'] = df['Salary']*0.1
df.head()
```

```python
#9.Sort employees by Salary in descending order.
df.sort_values("Salary",ascending=False)
```

```python
#10.Find the average salary of each department.
df.groupby("Department")['Salary'].mean()
```

```python
# Question 2
"""
data = {
    "Product": ["Laptop", "Mobile", "Laptop", "Headphones", "Mobile", "Laptop"],
    "Category": ["Electronics", "Electronics", "Electronics",
                 "Accessories", "Electronics", "Electronics"],
    "City": ["Nagpur", "Pune", "Nagpur", "Mumbai", "Pune", "Mumbai"],
    "Quantity": [2, 5, 3, 10, 4, 1],
    "Price": [50000, 20000, 50000, 2000, 20000, 50000]
}

1.Add a Total_Sales column
2.Find total sales of the entire store.
3.Find products with Quantity > 3.
4.Sort products by Total_Sales in descending order.
5.Find total sales by City
6.Find total sales by Product
7.Find the city with the highest total sales.
8.Use value_counts() to count how many times each product appears.
"""

data = {
    "Product": ["Laptop", "Mobile", "Laptop", "Headphones", "Mobile", "Laptop"],
    "Category": ["Electronics", "Electronics", "Electronics",
                 "Accessories", "Electronics", "Electronics"],
    "City": ["Nagpur", "Pune", "Nagpur", "Mumbai", "Pune", "Mumbai"],
    "Quantity": [2, 5, 3, 10, 4, 1],
    "Price": [50000, 20000, 50000, 2000, 20000, 50000]
}

df = pd.DataFrame(data)

#1.Add a Total_Sales column
df['Total_Sales'] = df['Quantity']*df['Price']
df.head()
```

```python
#2.Find total sales of the entire store.
print("Total Sales(Rs):",df['Total_Sales'].sum())
```

```python
#3.Find products with Quantity > 3.
df[df['Quantity']>3]
```

```python
#4.Sort products by Total_Sales in descending order.
df.sort_values("Total_Sales",ascending=False)
```

```python
#5.Find total sales by City
df.groupby("City")['Total_Sales'].sum()
```

```python
#6.Find total sales by Product
df.groupby("Product")['Total_Sales'].sum()
```

```python
#7.Find the city with the highest total sales.
df.groupby("City")['Total_Sales'].sum().idxmax()
```

```python
#8.Use value_counts() to count how many times each product appears.
df['Product'].value_counts()
```
