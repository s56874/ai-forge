```python
# Importing Seaborn Libirary
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
# 1.countplot

df = sns.load_dataset("tips")
df.head()
```

```python
sns.countplot(data=df, x="day")
plt.title("Count Plot")
plt.show()
```

```python
sns.countplot(data=df, x="day", hue="sex")
plt.title("Gender Wise Days")
plt.show()
```

```python
# 4.barplot
sns.barplot(data=df, x="day", y="total_bill")
plt.title("Bar Plot")
plt.show()
```

```python
sns.barplot(data=df, x='day', y='total_bill',hue='sex')
plt.title("Bar Plot")
plt.show()
```

```python
# scatterplot
plt.scatter(data=df, x='total_bill',y='tip')
plt.title("Scatter Plot")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex"
)
plt.title("Scatter Plot")
plt.show()
```

```python
# lineplot
sns.lineplot(data=df, x='total_bill',y='tip')
plt.show()
```

```python
# histplot
sns.histplot(data=df, x='total_bill')
plt.show()
```

```python
sns.histplot(data=df, x='total_bill',hue='sex')
plt.show()
```

```python
# Kde Plot
sns.kdeplot(
    data=df,
    x="total_bill"
)

plt.show()
```

```python
sns.kdeplot(
    data=df,
    x="total_bill",
    fill=True
)

plt.show()
```

```python
# Violinplot
sns.violinplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.show()
```

```python
# Heatmap
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.show()
```

```python
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)
```

```python
# pairplot
sns.pairplot(df)
plt.show()
```

```python
sns.pairplot(
    df,
    hue='sex',
    vars=['total_bill', 'tip']

)

plt.show()
```

```python
# jointplot
sns.jointplot(data=df, x='total_bill',y='tip')
plt.show()
```

```python
# regplot
sns.regplot(data=df, x='total_bill', y='tip')
plt.show()
```

```python
# catplot
sns.catplot(data=df, x='day',y='total_bill',kind='box')
plt.show()
```

```python
# question 1
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")
df.head()
```

```python
# 1.Which passenger class (class) had the highest survival rate, and how did survival differ between males and females?
sns.barplot(data=df, x='class',y='survived',hue='sex')
plt.title("Survival Rate by Passenger Class and Gender")
plt.show()
#  Ans- The First Class had the Highest Survival Rate and follwing graph show the female having highest survival rate than male acrros the passanger classes.
```

```python
#2.Which passenger class had the highest number of passengers?
sns.countplot(data=df,x='class')
plt.title("Number of Passangers of Class")
plt.show()
# Ans- The Third Class Passangrs had Highest no of passengers
```

```python
# How many male and female passengers were there in each passenger class?
sns.countplot(data=df, x='class',hue='sex')
plt.title("Number of Male and Female Passengers in Each Class")
plt.show()
# Ans-Following graph trends show the no of each class of male and female
```

```python
# Which age group had the highest number of passengers?
def age_group(age):
  if age <= 18:
    return "child"
  elif age <= 60:
    return 'Adult'
  else :
    return 'Senior'

df['Age_Group']  = df['age'].apply(age_group)

sns.countplot(data=df,x='class',hue='Age_Group')
plt.title("Number of Passengers by Age Group")
plt.show()

# Ans- Adult age group had the highest number of passengers
```

```python
# Which passenger class had the highest survival rate

sns.barplot(data=df, x='class', y='survived')
plt.title("Survival Rate by Passanger Class")
plt.show()
# Ans-# First Passangers class had the highest survival rate
```
