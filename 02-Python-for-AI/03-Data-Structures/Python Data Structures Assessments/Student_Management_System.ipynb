```python
from ast import Delete
student_list = []

while True:
    print("\n===== Student Management System =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update student")
    print("5. Count Total No Of Student")
    print("6. Delete Student")
    print("7. Exit")

    try:
        choice = int(input("Enter Your Choice: "))
    except ValueError:
        print("Please enter a valid number.")
        continue

    if choice == 1:
        name = input("Enter Student Name: ")
        if name in student_list:
          print("Student Already Exists")
        else:
         student_list.append(name)
         print("Student Added Successfully.")

    elif choice == 2:
        if not student_list:
            print("No students available.")
        else:
            print("\nStudent List:")
            for index, student in enumerate(student_list, start=1):
                print(f"{index}. {student}")

    elif choice == 3:
        name = input("Enter Student Name to Search: ")

        if name in student_list:

            print(f"{name} Found at System .")
        else:
            print("Student Not Found in System.")

    elif choice == 4:
        update = input("Enter Student Name to Update:")
        if update in student_list:
          index = student_list.index(update)
          new_name = input("Enter New Name:")
          if new_name not in student_list:
             student_list[index] = new_name
             print("Updated Successfully")
          else:
            print("Student Already exists.")
        else:
          print("Student Not Found in System")

    elif choice == 5:
      total_student = len(student_list)
      print("Total No of Student in System is",total_student)

    elif choice == 6:
      delete = input("Enter Name to Delete:")
      if delete in student_list:
        student_list.remove(delete)
        print("Deleted Successfully")
      else:
        print("Student not Found in System")

    elif choice == 7:
      print("===== Exiting Student Management System =====\nThank you for using our application!")
      break;
    else:
        print("Invalid Choice. Please enter a number between 1 and 6.")
```
