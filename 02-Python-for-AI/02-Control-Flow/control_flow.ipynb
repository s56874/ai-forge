```python
#  In this file we cover the following topics
#1. if Statement
#2. if-else
#3. if-elif-else
#4. Nested if
#5. for Loop
#6. while Loop
#7. Nested Loops
#8. break
#9. continue
#10. pass

#1. If Statement
marks = 90
if marks >= 85:
  print("Pass")
print("="*30)

#2. If-else statement
number = 7
if number % 2 == 0:
  print("Number is Even")
else:
  print("Number id Odd")
print("="*30)

#3. if-elif condition
signal = 'Yellow'
if signal == 'Red':
  print("Stop")
elif signal == 'Yellow':
  print("Great Ready")
elif signal == 'Green':
  print("Go")
else:
  print("Invalid Signal")
print("="*30)


#4. Nested-if condition
age = 20
has_voter_id = True
if age >= 18:
  if has_voter_id == True:
    print("Eligible for Voting")
  else:
    print("Required Voter id")
else:
  print("You are not eligible for Voting")
print("="*30)
```

```python
# 5.For Loop
for i in range(11):
  print(i)
print("="*30)

name = 'Samarth'
for ch in name:
  print(ch)
print("="*30)

for i in range(0,11,2):
  print(i)
print("="*30)

for i in range(1,6):
  print("*"*i)
print("="*30)

#6.While Loop
number = 1
while number <= 5:
  print(number)
  number += 1
print("="*30)

count = 5
while count >= 1:
  print(count)
  count -= 1
print("="*30)

# 7.Nestd Loop
for i in range(3):
  for j in range(3):
    print(i,j)
print("="*30)

#8.Break
for i in range(5):
  if i == 3:
    break;
  print(i)
print("="*30)

#9.Continue
for i in range(5):
  if i == 3:
    continue;
  print(i)
print("="*30)

#10.Pass statement

count = 1

while count <= 5:
    if count == 3:
        pass

    print(count)
    count += 1



```

```python
# Practice Question
 #1.Write a program that:
   #Checks whether the user has entered the correct PIN.
   #If the PIN is correct, check whether the account balance is sufficient.
   #Print "Transaction Successful" or "Insufficient Balance"

pin = 7555
balance = 50000

pin = int(input("Enter Your PIN:"))
if pin == 7555:
  amount = int(input("Enter Amount:"))
  if amount <= balance:
    print("Transaction Successful")
    remeaning_balance = balance - amount
    print("Remeaning Balance is",remeaning_balance)
  else:
    print("Insufficient Balance")
else:
  print("Invalid Pin")
```

```python
#4. Multiplication Table when our valye become 50 stop the loop

number = int(input("Enter a Number:"))
for i in range(1,11):
  result = number * i
  if result > 50:
    break;
  print(number,"*",i,"=",result)
```

```python
#Create a list of 5 student marks. Find and print:

 #Total marks
 #Average mark

marks = [85, 30, 76, 45, 20]
total = 0
count = 0
for mark in marks:
  if mark < 35:
    print("Marks:",mark,"Fail")
  else:
    print("Marks:",mark,"Pass")

    total += mark
    count += 1
average = total / count
print("Total Passing Marks:",total)
print("Average marks:",average)
```
