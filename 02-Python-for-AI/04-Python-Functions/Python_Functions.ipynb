```python
# Write Program To Find Given Number is Even Or Odd

def even_or_odd(a):
  if a % 2 == 0:
    return 'Even'
  else:
    return 'Odd'
a = int(input("Enter A Number:"))
print(even_or_odd(a))
```

```python
# Write A a Function Check Number is Positive ,Negative and Zero

def number_cheaker(num):
  if num > 0:
    return  'Positive'
  elif num < 0 :
    return 'Negative'
  else:
    return 'zero'

num = int(input("Enter a Number:"))
print(number_cheaker(num))
```

```python

# Write a function that returns the larger number

def largest(a,b,c):
  if a > b and a > c:
    return a
  elif b > a and b > c:
    return b
  else:
    return c

a = int(input("Enter First Number:"))
b = int(input("Enter Second Number:"))
c = int(input("Enter Third Number:"))
print("Largest Number is:",largest(a,b,c))
```

```python
# Write a function to check whether a number is Prime or Not Prime.
def prime(num):
  if num <= 1:
    return 'Not Prime'

  for i in range(2,num):
      if num % i == 0:
        return 'Not Prinme'


  return 'Prime'

num = int(input("Enter a number:"))
print(prime(num))
```

```python
# Write a prgram to find Vowles in the string

def count_vowles(string):
  vowles = 'aeiouAEIOU'
  count = 0
  for char in string:
    if char in vowles:
      count += 1
  return count

string = input("Enter a String:")
print(count_vowles(string))
```

```python
# Reverse a String
def reverse_string(string):
  reverse = ""
  for ch in string:
   reverse = ch + reverse
  return reverse

string = input("Enter a String to Reverse:")
print(reverse_string(string))
```

```python
# Reverse a String using builtin function
def reverse_strinng(string):
  return string[::-1]
string ='python'
print(reverse_string(string))
```

```python
# Count Digits
def count_digit(digit):
  count = 0
  while digit != 0:
    digit //= 10
    count += 1
  return count

digit = int(input("Enter A Number:"))
print(count_digit(digit))
```

```python
# Sum of Digits Using Function
def sum_of_digit(digit):
  sum = 0
  for num in str(digit):
    sum += int(num)
  return sum
digit = int(input("Enter a Number:"))
print("Sum of Digit:",sum_of_digit(digit))
```

```python
# Find factrotrial using Function
def factrotrial(num):
  result = 1
  for digit in range(1,num+1):
    result *= digit
  return result
num = int(input("Enter a Number:"))
print(factrotrial(num))
```
