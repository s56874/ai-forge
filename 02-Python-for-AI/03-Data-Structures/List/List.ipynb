```python
# Study of List

# 1.Cretae List(Creating List)
numbers = [10,20,30,40,50]
print(numbers)

# 2.Empty list
l = []
print(l)

# 3.Acces Element
print(numbers[0]) # First Element
print(numbers[-1]) #Last Element

# 4.Slicing
print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[::-1]) # Reverse

# 5.update Element
numbers[1]  = 200
print(numbers)

# 6.Appending Element
numbers.append(60)
print(numbers)

# 7.Inserting Element
numbers.insert(2,80)
print(numbers)

# 8.Extend Element
numbers.extend([70,80,90])
print(numbers)

# 9.Remove Element
numbers.remove(10)
print(numbers)

# 10. Pop (Delete) Operation
numbers.pop()
print(numbers)

numbers.pop(0)
print(numbers)

# 11.Delete Element
del numbers[0]
print(numbers)

# 12.Length of List
print(len(numbers))

# 13.Membership
print(30 in numbers)
print(200 in numbers)

# 14. Counting Numbers
l = [1,1,2,3,2,1,3,5]
print(l.count(2))

# 15.Indexing
print(l.index(3))

# 16.Sorting Element
l = [50,10,30,20]
l.sort()
print(l)

# 17.Descending Order
l.sort(reverse=True)
print(l)

# 18.Reversing Element
l.reverse()
print(l)

# 19.Copy List
copy_list = l.copy()
print(copy_list)

# 20.Clear List
copy_list.clear()
print(copy_list)

# 21. Maximum Element
print(max(l))

# 22.Minimum Element
print(min(l))

# 23.Sum of all Element
print(sum(l))

# 24. Repeatation
print(l*2)

# 25.Loop
for i in l:
  print(i)

# 26.List Adding
l1 = [1,2,3,4]
l2 = [5,6,7,8,]
print(l1+l2)

# 27. Nested List
matrix = [
    [1,2],
    [3,4]
]
print(matrix[0][1])

# 28. List Comprehension
square = [x*x for x in range(1,6)]
print(square)

# 29.Remove Duplicate
l = [1,2,2,3,3,4,5,5]
unique = list(set(l))
print(unique)

# 30.Input in List
l = []
n = int(input("How Many Element Add You Want:"))
for i in range(n):
  num = int(input("Enter Number:"))
  l.append(num)
  print(l)
```

```python
# Question 2
"""
2.Student Record Management

Scenario: A school stores students' marks.

Task:

1.Store marks in a list.
2.Find the highest and lowest marks.
3.Calculate the average without using sum().
4.Count how many students passed (marks ≥ 35).
4.Display the names/marks of students who scored above average.
"""
#1. Store makrks in List
marks = [88,55,23,55,34,90,65,44]
print(marks)


#2.Find Highest and Lowest Marks
highest_marks =  marks[0]
lowest_marks  = marks[0]
for mark in marks:
  if mark > highest_marks:
      highest_marks = mark
  if mark < lowest_marks:
     lowest_marks = mark
print(f"Highest Marks:",highest_marks)
print(f"Lowest Marks:",lowest_marks)

#3.Calculate the average without using sum().
total = 0
for mark in marks:
  total += mark
average = total / len(marks)
print("Average Marks is:",average)

#4.Count how many students passed (marks ≥ 35).
passed_student = 0
failed_student = 0
for mark in marks:
  if mark >= 35:
    passed_student += 1
  else:
    failed_student += 1
print("Passed Student:",passed_student)
print("Failed student:",failed_student)

#5.Display the marks of students who scored above average.
average_marks = [mark for mark in marks if mark > average]
print("Marks of Student which is score above average marks :",average_marks)

```

```python
# Queation 4
"""
Inventory Management

Scenario: A warehouse keeps track of product stock.

Task:
1.Store product quantities in a list.
2.Update stock after a sale.
3.Remove products with zero stock.
4.Find the product with the highest quantity.
5.Display the updated inventory.

 """

 #1. Store product quantities in a list
print("Inventory Management")
print("="*30)
products = ["Laptops", "Keyboards", "Monitors", "Mice", "Headphones"]
stock = [15, 50, 8, 42, 24]
print("Avilable Stocks:")
print("="*30)
print("Initial Inventory:",products)
print("Initial Stock:",stock)

#2.Update stock after a sale.
# Suppose we say out of 8 monitors sold
sold_monitors = 8
stock[2] -= sold_monitors
print("="*30)
print("Updated Inavntry:",products)
print("Updated Stock:",stock)

#3.Remove products with zero stock.
products.pop(2)
stock.pop(2)
print("="*30)
print("Updataed Products:",products)
print("Updated Stock:",stock)

#4.Find the product with the highest quantity.
highest_quanity_product = products[0]
highest_quanity_stock = stock[0]
for i in range(len(stock)):
  if stock[i] > highest_quanity_stock:
    highest_quanity_stock = stock[i]
    highest_quanity_product = products[i]
print("="*30)
print("Highest Quantity Product:",highest_quanity_product)
print("Highest Quantity Stock:",highest_quanity_stock)

#5.Display the updated inventory
print("="*30)
print("Final Inventory:",products)
print("Final Stock:",stock)
```

```python
# Question 5
"""
Sales Report

Scenario: A company records daily sales.

Task:
1.Store 7 days of sales in a list.
2.Calculate the total sales without using sum().
3.Find the highest and lowest sales.
4.Calculate the average sales.
5.Display all sales values greater than the average
"""

#1.Store 7 days of sales in a list.
sales_data = [10000,50000,100000,500000,10000000,45000000,25000000]
print(sales_data)
print("="*30)

#2.Calculate the total sales without using sum()
total_sale = 0
for i in sales_data:
  total_sale += i
print("Total Sale:",total_sale,"Rs")
print("="*30)

#3.Find the highest and lowest sales.
highest_sale = sales_data[0]
lowest_sale  = sales_data[0]
for i in sales_data:
   if i > highest_sale:
    highest_sale = i
    if i < lowest_sale:
      lowest_sale = i
print("Highest sale is:",highest_sale,"Rs")
print("Lowest sale is:",lowest_sale,"Rs")
print("="*30)


#4.Calculate the average sales.
average_sale = total_sale / len(sales_data)
print("Average sale is:",average_sale,"Rs")
print("="*30)

#5.Display all sales values greater than the average
for i in sales_data:
   if i > average_sale:
    print("Sales is above than average:",i,"Rs")
```
