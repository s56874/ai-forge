```python
# ======Contact Book Management system =====



contacts = {
    "Samarth": "9876543210",
    "Rahul": "9123456789",
    "Amit": "9012345678",
    "Priya": "9988776655",
    "Sneha": "9876501234"
}


while True:
  print("\n===== Contact Management System =====")
  print("1.View Contact")
  print("2.Add Contact")
  print("3.Update Contact")
  print("4.Delete Contact")
  print("5.Search Contacts")
  print("6.Exit")
  print("="*30)


  try:

      choice = int(input("Enter your Choice:"))
  except ValueError:
       print("Invalid Input! Please Enter Valid Number.")
       continue


  #1.To View all contact in the dictionary
  if choice == 1:
    print("\n=====View Contact=====")
    for name, phone in contacts.items():
      print(f"{name}: {phone}")

  #2. Add contact number in the book
  elif choice == 2:
    print("\n=====Add Contact=====")
    name = input("Enter Name:" )
    phone = input("Enter phone Number:")
    contacts[name] = phone
    print("Contct is added successfully")


  #3.Update Contact
  elif choice == 3:
      print("\n=====Update Contact=====")
      name = input("Enter Name to Update:")
      if name not in contacts:
        print("Contact Nois not Avilable Contact List.")
      else:
        phone = input("Enter New Phone Number:")
        contacts[name] = phone
        print("Contacts is Updated Successfully")


   #4. Delete Contact
  elif choice == 4:
    name = input("Enter Name to Delete:")
    if name in contacts:
      contacts.pop(name)
      print("Contact Deleted Successfully")
    else:
      print("Contact Not Avilable in Contact List")


   #5.Search Contacts
  elif choice == 5:
    print("======Search Contacts=====")
    print("1.Search by Name:")
    print("2.Search by Number:")
    search_choice = int(input("Enter Your Choice:"))
    if search_choice == 1:
      name = input("Enter Name to Search:")
      if name in contacts:
        print(f"{name}:{contacts[name]}")
      else:
        print("Contacts Not Avilable in Contacts List")
    elif search_choice == 2:
      phone = input("Enter Phone Number to Search:")
      for name, phone_number in contacts.items():
        if phone_number == phone:
          print(f"{name}:{phone_number}")
          break;
    else:
      print("Invalid Input! Please Enter Valid Input.")





  #6. Exit the System
  elif choice == 6:
    print("=====Exiting System=====")
    print("Thanks For Choosing Us")
    break;
  else:
    print("Invalid Input! Please Enter Valid Input")
```
