```python
# Study Of Touple

# 1.Creating Touple
t = (10,20,30,40,50)
print("Touple:",t)

# 2.Indexing
print("First Element:",t[0])
print("Last element:",t[-1])

# 3.Slicing
print("Slice:",t[1:4])

# 4.Length
print("Length:",len(t))

# 5.Membership
print(30 in t)

# 6.Counting Numbers
print("Count of 20:",t.count(20))

# 7.Indexing
print("Index of 30",t.index(30))

# 8.Maximun Number
print("Maximum:",max(t))

# 9.Minimun Number
print("Minimun:",min(t))

# 10. Sum of Numbers In touple
print("Total Sum:",sum(t))

# 11. Loop Condition
print("Loop:")
for i in t:
  print(i)

# 12. Conversion Touple to list
l = list(t)
print("List",l)

# 13. Conversion List to Touple
t = tuple(l)
print("Tuple:",t)
```

```python
# Practice Question
# Question 1

"""
A university stores each student's record as a tuple.

student = ("S101", "Samarth", "CSE-DS", 7.50, "Nagpur")
Tasks:
1.Print the student ID.
2.Print the student's name and branch.
3.Check if the student belongs to "CSE-DS
4.Print the total number of fields.
"""
#1.Print the student ID.
student = ("S101", "Samarth", "CSE-DS", 7.50, "Nagpur")
print("Student Id:",student[0])
print("="*30)

#2.Print the student's name and branch.
print("Student Name:",student[1])
print("Student Branch:",student[2])
print("="*30)

#3.Check if the student belongs to "CSE-DS
if student[2] == 'CSE-DS':
  print("Student Belongs to CSE-DS")
else:
  print("Student does not Belong to CSE-DS")
print("="*30)
#4.Print the total number of fields.
print("Total no of Fields:", len(student))
print("="*30)

print("-----Student Summary-----")
print("Student Id:",student[0])
print("Student Name:",student[1])
print("Branch:",student[2])
print("CGPA:",student[3])
print("City:",student[4])
print("="*30)



```

```python
# Question 2
"""
2.Hospital Patient Record

A hospital stores patient details in a tuple.

patient = ("P205", "Amit", 28, "Fever", "Ward-3")
Tasks
1.Print all patient details.
2.Print the patient's age.
3.Check whether the patient is admitted to "Ward-3".
4.Print the first and last elements.
"""

#1.Print all patient details
patient = ("P205", "Amit", 28, "Fever", "Ward-3")
print("Patient Details:",patient)
print("="*30)

#2.Print the patient's age
print("Patient Age:",patient[2])
print("="*30)

#3..Check whether the patient is admitted to "Ward-3"
if patient[4] == 'Ward-3':
  print("Patient Admitted to Ward-3")
else:
  print("Patient not Admitted in ward-3")
print("="*30)

#4.Print the first and last elements
print("First Element:",patient[0])
print("Last Element:",patient[4])
print("="*30)

print("-----Patient Summary-----")
print("Patient Id:",patient[0])
print("Patient Name:",patient[1])
print("Age:",patient[2])
print("Disese:",patient[3])
print("Ward:",patient[4])
print("="*30)
```
