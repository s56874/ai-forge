```python
# Study of Set

# 1.Creating Set
s = {10,20,30,40}
print(s)

# 2. Adding Element
s.add(50)
print("After Add:",s)

# 3.Update Element in Set
s.update([50,60])
print("After Update:",s)

# 4.Removing Element
s.remove(20)
print("After Remove:",s)

# 5.Discarding Element
s.discard(50)
print("After Discard:",s)

# 6. Pop Element
print("Pooped:",s.pop())
print(s)

# 7.Membership of Element
print(30 in s)

# 8.Union,Intersection
a = {1,2,3}
b = {4,5,6}
print("Union:",a.union(b))
print("Intersection:",a.intersection(b))
print("Difference:",a.difference(b))
print("Symentric Difference:",a.symmetric_difference(b))
```

```python
#Prctice Question 1
"""
1. Removing Duplicate Records

A company's employee database contains duplicate employee IDs.

employee_ids = [101, 102, 103, 101, 104, 105, 102, 106]
Tasks:
1.Convert the list into a set.
2.Print the unique employee IDs.
3.Print the total number of unique employees.
4.Display the IDs in sorted order.
"""

#1.Convert the list into a set.
employee_ids = [101, 102, 103, 101, 104, 105, 102, 106]
print(set(employee_ids))
print("="*30)

#2.Print the unique employee IDs.
print("Unique Employee Ids:",set(employee_ids))
print("="*30)

#3.Print the total number of unique employees.
print("Total No of Unique Employee Ids:",len(set(employee_ids)))
print("="*30)

#4.Display the IDs in sorted order.
print("Sorted Employe Ids:",sorted(set(employee_ids)))
print("="*30)
```

```python
# Practice Question 2
"""
2. Unique Email Management

A website receives user email registrations, but some users register multiple times.

emails = [
    "amit@gmail.com",
    "rahul@gmail.com",
    "amit@gmail.com",
    "neha@gmail.com",
    "rahul@gmail.com",
    "samarth@gmail.com"
]
Tasks:
1.Store only unique email addresses using a set.
2.Print all unique email addresses.
3.Count the total number of registered users.
4.Check whether "samarth@gmail.com" is registered.
"""

#1.Store only unique email addresses using a set.
emails = [
    "amit@gmail.com",
    "rahul@gmail.com",
    "amit@gmail.com",
    "neha@gmail.com",
    "rahul@gmail.com",
    "samarth@gmail.com"
]
print("Unique Email Address:",set(emails))
print("="*30)

#2.Print all unique email addresses.
for email in set(emails):
  print(email)
print("="*30)

#3.Count the total number of registered users.
print("Total No Of Registered User:",len(set(emails)))
print("="*30)

#4.Check whether "samarth@gmail.com" is registered.
print("samarth@gmail.com" in  emails)
print("="*30)
```

```python
#Preactice Quesion 3
"""
3.Student Enrollment System

Students have enrolled in two elective courses.

python_course = {
    "Amit",
    "Rahul",
    "Neha",
    "Samarth"
}

ai_course = {
    "Rahul",
    "Priya",
    "Samarth",
    "Rohan"
}
Tasks:
1.Find students enrolled in both courses.
2.Find all enrolled students.
3.Find students enrolled only in the Python course.
4.Find students enrolled only in the AI course.
"""

#1.Find students enrolled in both courses.

python_course = {
    "Amit",
    "Rahul",
    "Neha",
    "Samarth"
}

ai_course = {
    "Rahul",
    "Priya",
    "Samarth",
    "Rohan"
}
print("Student Enrolled in Both Classes:",python_course.intersection(ai_course))
print("="*30)

#2.Find all enrolled students.
print("All Enrolled Student:",python_course.union(ai_course))
print("="*30)

#3.Find students enrolled only in the Python course.
print("Student Enrolled in only Python Classes:",python_course.difference(ai_course))
print("="*30)

#4.Find students enrolled only in the AI course.
print("Student Enrolled in Only Ai Classes:",ai_course.difference(python_course))
print("="*30)
```

```python
#Practice Question 4

"""
4. Product Category Management

An online store receives product categories from two suppliers.

supplier1 = {
    "Electronics",
    "Books",
    "Clothing",
    "Sports"
}

supplier2 = {
    "Books",
    "Furniture",
    "Electronics",
    "Toys"
}
Tasks:
1.Find all available categories.
2.Find common categories.
3.Find categories available only from Supplier 1.
4.Find categories available only from Supplier 2.
"""

#1.Find all available categories.

supplier1 = {
    "Electronics",
    "Books",
    "Clothing",
    "Sports"
}

supplier2 = {
    "Books",
    "Furniture",
    "Electronics",
    "Toys"
}
print("Total Avilable Categories:",supplier1.union(supplier2))
print("="*30)

#2.Find common categories.
print("Common Categories :",supplier1.intersection(supplier2))
print("="*30)

#3.Find categories available only from Supplier 1.
print("Catergories Avilable Only Supplier1:",supplier1.difference(supplier2))
print("="*30)

#4.Find categories available only from Supplier 2.
print("Catergories Avilable Only Suppliers2:",supplier2.difference(supplier1))
print("="*30)
```

```python
#Practice Question 5
"""
5.Data Cleaning and Preprocessing

A data analyst receives customer IDs containing duplicate values.

customer_ids = [
    1001, 1002, 1003, 1002,
    1004, 1001, 1005, 1006,
    1003
]
Tasks:
1.Remove duplicate customer IDs using a set.
2.Print the cleaned customer IDs.
3.Count the number of unique customers.
4.Check whether customer ID 1007 exists.
"""

#1.Remove duplicate customer IDs using a set.
customer_ids = [
    1001, 1002, 1003, 1002,
    1004, 1001, 1005, 1006,
    1003
]
print("Unique Customer IDs:",set(customer_ids))
print("="*30)

#2.Print the cleaned customer IDs.
for id in set(customer_ids):
  print(id)
print("="*30)

#3.Count the number of unique customers.
print("Total No Of Unique Customers Ids:",len(set(customer_ids)))
print("="*30)

#4.Check whether customer ID 1007 exists.
if 1007 in set(customer_ids):
  print("1007 Customer Id Exists")
else:
  print("1007 Customer  Id not Exists")
print("="*30)
```
