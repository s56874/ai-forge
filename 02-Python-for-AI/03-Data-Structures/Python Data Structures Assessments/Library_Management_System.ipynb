```python
# =====Library Management System
book_list =[]


while True:
  print("\n===== Library Management System =====")
  print("1. Add Book")
  print("2. Delete Book")
  print("3. Update Book")
  print("4. Display Book")
  print("5. Search Book")
  print("6. Count Total Records")
  print("7. Exit")
  print("="*30)



  choice = int(input("Enter Your Choice: "))




  # 1. Add bookin the List
  if choice == 1:
    book = input("\nEnter Book Name:")
    if book in book_list:
      print("\nBook Already Avilable in Sytem")
    else:
       book_list.append(book)
       print("\nBook Added Successfully.")

  # 2. Delete book in the list
  elif choice == 2:
    book = input("\nEnter Book Name:")
    if book in book_list:
      book_list.remove(book)
      print("\n Record Deleted Succesfully")
    else:
      print("\nBook not Found in System")

  #3.  Update Book Name
  elif choice == 3:
   book = input("\nEnter Book Name:")
   if book in book_list:
    index = book_list.index(book)
    new_book = input("\nEnter New Book Name:")
    book_list[index] = new_book
    print("\nBook Updated Successfully")
   else:
    print("Book Not Found in System")


  #4. Display Book

  elif choice == 4:
    if not  book_list:
      print("\nNo Book Avilable System")
    else:
      print("\nRecord Avilabe in System")
      for index , book in enumerate(book_list, start=1):
        print(index,":",book)

  #5. Searching Book in the List
  elif choice == 5:
    book = input("Enter Book Name to Search:")
    if book in book_list:
      index = book_list.index(book)
      print(f"{book} Found at Position {index +1}")
    else:
      print("Book Not Found in System")


  # 6. Find Total Records In The System
  elif choice == 6:
    if  not  book_list:
      print("No Found Books in the System")
    else:
      print("total No of Books Available in system is",len(book_list))


  #7. Exit System
  elif choice == 7:
    print("=====Exiting=====")
    print("Thanks for Choosing Library Management System")
    break;
  else:
    print("Invalid Input")



```
