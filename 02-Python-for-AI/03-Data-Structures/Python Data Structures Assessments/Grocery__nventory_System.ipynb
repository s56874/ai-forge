```python
# =======Grocery Inventory System======
product = {
     "Rice": {
        "Category": "Grains",
        "Quantity": 20,
        "Price": 55.0
    }
}
categories = (
    "Grains",
    "Beverages",
    "Spices",
    "Cooking Oil",
    "Snacks"
)
history = []

while True:
  print("\n=====Grocery Inventory System======")
  print("1. View Categories")
  print("2. Add Product")
  print("3. View Products")
  print("4. Search Product")
  print("5. Update Quantity")
  print("6. Delete Product")
  print("7. Inventory Summary")
  print("8. Recently Added Products")
  print("9. Exit")
  print("="*30)


  # Taking Input from User
  try:
        choice = int(input("Enter Your Choice: "))
  except ValueError:
      print("Please enter a valid number.")
      continue


  #1.View Categories Avilabel In the system

  if choice== 1:
    print("======Categories Avilable In Inventory=====")
    for index, items in enumerate(categories,start=1):
      print(index,":",items)



  #2.Adding Product in the system (Inventory)
  elif choice == 2:
    print("======Adding Product=====")
    print("Enter Product Details:")
    name = input("Product Name:")
    if name in product:
      print("Product Already Exists.")

    else:
      category = input("Category:")
      quantity = int(input("Quantity:"))
      price = float(input("Price:"))
      product[name] = {
        "Category" : category,
        "Quantity" : quantity,
        "Price" : price
      }
      history.append(name)
      print("Product Added Succssfully.")



  #3.View Products Avilable in System
  elif choice == 3:
    print("====== Available Products ======")

    if not product:
        print("No Products Available.")
    else:
        for index, (name, details) in enumerate(product.items(), start=1):
            print(f"\nProduct {index}")
            print(f"Name     : {name}")
            print(f"Category : {details['Category']}")
            print(f"Quantity : {details['Quantity']}")
            print(f"Price    : ₹{details['Price']}")
            print("-" * 30)


  #4.Search Product in the System
  elif choice == 4:
    print("======Search Product======")
    name = input("Enter Product Name:")
    if name in product:
      print("Product Found",name)
      print(product[name])
    else:
      print("Product not Found  in System")


  #5.Update Quantity in the sytsem
  elif choice == 5:
    print("=======Update Inventory======")
    name = input("Enter Product Name:")
    if name in product:
      print("1.Update Category")
      print("2.Update Quantity")
      print("3.Update Price")
      try:
         choice = int(input("Enter Your Choice: "))
      except ValueError:
         print("Please enter a valid number.")
         continue
      if choice == 1:
        category = input("Enter New Category:")
        product[name]["Category"] = category
        print("Category Updated Successfully.")

      elif choice == 2:
        quantity = int(input("Enter New Quantity:"))
        product[name]["Quantity"] = quantity
        print("Quantity Updated Successfully.")

      elif choice == 3:
        price = float(input("Enter New Price:"))
        product[name]["Price"] = price
        print("Price Updated Successfully.")

      else:
        print("Invalid Input! Please Enter Valid Input")

    else:
      print("Product Not Found System")


  #6.Delete Product in the System
  elif choice == 6:
    print("======Delete Product======")
    name = input("Enter Product Name:")
    if name in product:
      product.pop(name)
      print("Product Deleted successfully.")
    else:
      print("Product Not Found.")

  #7.Inventory Summary of System
  elif choice == 7:
    print("======Inventory Summary=======")
    print("\nTotal Product",len(product))
    for name, details in product.items():
      print(f"\nProduct: {name}")
      print(f"Category: {details['Category']}")
      print(f"Quantity: {details['Quantity']}")
      print(f"Price: ₹{details['Price']}")
      print("-" * 30)
    total_quantity = 0
    total_price = 0
    for name, details in product.items():
      total_quantity += details['Quantity']
      total_price += details['Quantity'] * details['Price']
    print(f"\nTotal Quantity:{total_quantity}")
    print(f"Total Value of Inventory ₹:{total_price}")



  #8.Recently Added Products To Show the Users
  elif choice == 8:
    print("=======Recently Added Products========")

    if not history:
      print("No Product Added Recently.")
    else:
      print("\nRecently Added Product")
      for index, items in enumerate(history,start=1):
        print(index,":",items)


  #9.Exit the System
  elif choice == 9:
    print("=====Exixting System======")
    print("Thanks For Chooseing Our System.")
    break;

  else:
    print("Invalid Input Enter Valid Input.")
```
